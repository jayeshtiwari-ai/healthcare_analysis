```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, ConfusionMatrixDisplay

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

print("TensorFlow version:", tf.__version__)
```

```python
sentiment_df = pd.read_csv("synthetic_patient_sentiment.csv")
document_df = pd.read_csv(
    "synthetic_medical_document_classification.csv"
)

print("Sentiment dataset:", sentiment_df.shape)
print("Document dataset:", document_df.shape)

display(sentiment_df.head())
display(document_df.head())
```

```python
print("Sentiment classes:")
print(sentiment_df["sentiment"].value_counts())

print("\nDocument categories:")
print(document_df["category"].value_counts())

print("\nMissing values in sentiment dataset:")
print(sentiment_df.isnull().sum())

print("\nMissing values in document dataset:")
print(document_df.isnull().sum())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sentiment_df["sentiment"].value_counts().plot(
    kind="bar", ax=axes[0], title="Sentiment Distribution"
)

document_df["category"].value_counts().plot(
    kind="bar", ax=axes[1], title="Medical Category Distribution"
)

plt.tight_layout()
plt.show()
```

```python
def prepare_data(df, text_col, label_col, max_words=5000, max_len=60):
    df = df[[text_col, label_col]].dropna().copy()

    df[text_col] = df[text_col].astype(str).str.strip()
    df = df[df[text_col] != ""]

    # Encode class labels
    encoder = LabelEncoder()
    y = encoder.fit_transform(df[label_col])

    # Split before fitting the tokenizer to avoid vocabulary leakage
    X_train_text, X_test_text, y_train, y_test = train_test_split(
        df[text_col],
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    # Learn vocabulary only from training text
    tokenizer = Tokenizer(
        num_words=max_words,
        oov_token="<OOV>"
    )
    tokenizer.fit_on_texts(X_train_text)

    X_train = pad_sequences(
        tokenizer.texts_to_sequences(X_train_text),
        maxlen=max_len,
        padding="post",
        truncating="post"
    )

    X_test = pad_sequences(
        tokenizer.texts_to_sequences(X_test_text),
        maxlen=max_len,
        padding="post",
        truncating="post"
    )

    return (
        X_train, X_test,
        y_train, y_test,
        tokenizer, encoder
    )
```

```python
(
    X_train_sent,
    X_test_sent,
    y_train_sent,
    y_test_sent,
    sentiment_tokenizer,
    sentiment_encoder
) = prepare_data(
    sentiment_df,
    text_col="text",
    label_col="sentiment"
)

print("Training shape:", X_train_sent.shape)
print("Testing shape:", X_test_sent.shape)
print("Classes:", sentiment_encoder.classes_)
```

```python
def build_lstm_model(num_classes, max_words=5000, max_len=60):
    model = Sequential([
        Embedding(
            input_dim=max_words,
            output_dim=64,
            input_length=max_len
        ),
        LSTM(64),
        Dropout(0.3),
        Dense(32, activation="relu"),
        Dense(num_classes, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
sentiment_model = build_lstm_model(
    num_classes=len(sentiment_encoder.classes_)
)

sentiment_model.summary()
```

```python
history_sent = sentiment_model.fit(
    X_train_sent,
    y_train_sent,
    epochs=8,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history_sent.history["accuracy"], label="Training")
plt.plot(history_sent.history["val_accuracy"], label="Validation")
plt.title("Sentiment Model Accuracy")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_sent.history["loss"], label="Training")
plt.plot(history_sent.history["val_loss"], label="Validation")
plt.title("Sentiment Model Loss")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
sent_probs = sentiment_model.predict(X_test_sent)
sent_preds = np.argmax(sent_probs, axis=1)

print(classification_report(
    y_test_sent,
    sent_preds,
    labels=np.arange(len(sentiment_encoder.classes_)),
    target_names=sentiment_encoder.classes_,
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test_sent,
    sent_preds,
    labels=np.arange(len(sentiment_encoder.classes_)),
    display_labels=sentiment_encoder.classes_,
    xticks_rotation=45
)

plt.tight_layout()
plt.show()
```

```python
def predict_sentiment(text):
    sequence = sentiment_tokenizer.texts_to_sequences([text])

    padded = pad_sequences(
        sequence,
        maxlen=60,
        padding="post",
        truncating="post"
    )

    probabilities = sentiment_model.predict(padded, verbose=0)[0]
    index = np.argmax(probabilities)

    print("Predicted sentiment:", sentiment_encoder.classes_[index])
    print(f"Model score: {probabilities[index]:.2%}")

predict_sentiment(
    "The doctor was kind and explained everything clearly."
)
```

```python
(
    X_train_doc,
    X_test_doc,
    y_train_doc,
    y_test_doc,
    document_tokenizer,
    document_encoder
) = prepare_data(
    document_df,
    text_col="text",
    label_col="category"
)

print("Training shape:", X_train_doc.shape)
print("Testing shape:", X_test_doc.shape)
print("Categories:", document_encoder.classes_)
```

```python
document_model = build_lstm_model(
    num_classes=len(document_encoder.classes_)
)

document_model.summary()
```

```python
history_doc = document_model.fit(
    X_train_doc,
    y_train_doc,
    epochs=8,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history_doc.history["accuracy"], label="Training")
plt.plot(history_doc.history["val_accuracy"], label="Validation")
plt.title("Document Model Accuracy")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_doc.history["loss"], label="Training")
plt.plot(history_doc.history["val_loss"], label="Validation")
plt.title("Document Model Loss")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
doc_probs = document_model.predict(X_test_doc)
doc_preds = np.argmax(doc_probs, axis=1)

print(classification_report(
    y_test_doc,
    doc_preds,
    labels=np.arange(len(document_encoder.classes_)),
    target_names=document_encoder.classes_,
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test_doc,
    doc_preds,
    labels=np.arange(len(document_encoder.classes_)),
    display_labels=document_encoder.classes_,
    xticks_rotation=45
)

plt.tight_layout()
plt.show()
```

```python
def classify_document(text):
    sequence = document_tokenizer.texts_to_sequences([text])

    padded = pad_sequences(
        sequence,
        maxlen=60,
        padding="post",
        truncating="post"
    )

    probabilities = document_model.predict(padded, verbose=0)[0]
    index = np.argmax(probabilities)

    print("Predicted category:", document_encoder.classes_[index])
    print(f"Model score: {probabilities[index]:.2%}")

classify_document(
    "This study examines asthma symptoms, lung function, and respiratory infections."
)
```
